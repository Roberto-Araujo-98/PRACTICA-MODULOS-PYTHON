```python
import nltk
from nltk.tokenize import word_tokenize, TweetTokenizer
from nltk.corpus import stopwords
```

```python
nltk.download("punkt_tab")
nltk.download("stopwords")
```

```python
stop_words=stopwords.words("spanish")
stop_words
```

```python
texto = "Hola a todos ¿Cómo están? Hoy aprenderemos NLTK"
```

```python
tokens = TweetTokenizer().tokenize(texto.lower())
print(tokens)
```

```python
texto_limpio = [t for t in tokens if t.isalpha() and not t in stop_words]
texto_limpio
```

```python
[ t for t  in word_tokenize(texto) if t.isalpha() and not t in stopwords.words("spanish") ]
```
