```python
nlp = spacy.load("es_core_news_sm")
```

```python
texto2 = "El instructor Jesús Romero Villanueva, dictará un curso en SENATI en la ciudad de Arequipa"
```

```python
tokens =nlp(texto2)
```

```python
for t in tokens.ents: #RECONOCIMIENDTO DE INDETIFICADORES
    print(t)
```

## Procesamiento del texto

```python
tokens = nlp(texto2.lower()) #procesamos nuestro texto
print([t for t in tokens])
```

```python
for t in tokens: #FOR NATIVO
    print([t])
```

## Lematización

```python
print([(t.text, t.lemma_) for t in tokens]) #lematización
```

```python
for t in tokens:
    print([t.text, t.lemma_])  #FOR NATIVO
```

## Etiquetado Gramatical ( POS TAGGING)

```python
print([(t.text, t.tag_) for t in tokens]) #etiquetado gramatical (POS TAGGING)
```

```python
for t in tokens:
    print([t.text, t.tag_])    #FOR NATIVO
```
