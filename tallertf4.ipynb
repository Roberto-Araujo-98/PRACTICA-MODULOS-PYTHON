```python
import tensorflow as tf
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
```

```python
iris = load_iris()
x = iris.data   #x debe ser 2d
y = iris.target #y debe ser 1d
```

```python
print("Dato",x[0])
print("Etiqueta",y[0])
print("Nombre Etiqueta",iris.target_names[y[149]])
print("Caràcteristicas del dataset", iris.feature_names)
print("cant. de caracterìsticas (Para input layer:)", len(iris.feature_names))
print("targets:", len(iris.target_names))
print(" cant. de etiquetas (Para output layer):", len(iris.target_names))
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state= 42)
```

```python
model= tf.keras.models.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy')
```

```python
model.fit(x_train, y_train, epochs = 100)
```

```python
print('---Predicción Individual---')
muestra= x_test[0].reshape(1,4)
print('Muestra seleccionada: ', muestra)
etiqueta= y_test[0]
print('Etiqueta de la muestra: ', etiqueta)
prediccion= model.predict(muestra)
print("Probabilidades reales de nuestra salida",prediccion)
print("Etiqueta con la probabilidad máxima",prediccion.argmax(1))
```

```python
print('---Predicción Total---')
predicciones= model.predict(x_test).argmax(1)
print('Predicciones')
print(predicciones)
print('Etiquetas reales')
print(y_test)
```
