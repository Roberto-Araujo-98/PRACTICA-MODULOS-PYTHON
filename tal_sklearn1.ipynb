```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier #Clasificación multiclase
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
```

```python
iris= load_iris() #cargamos el dataset
x= iris.data #obtenemos los datos (x)
y= iris.target #obtenemos las etiquetas (y)
```

```python
print(x.shape) #obtenemos la forma de x (2D)
print(x) 
```

```python
print(y.shape) #forma de y (1D)
print(y)
```

```python
#---ANÁLISIS EXPLORATORIO DEL DATASET---
frame = pd.DataFrame(x,y,columns= iris.feature_names) #CREAMOS EL DATAFRAME CON PANDAS
frame['etiqueta']= iris.target #AGREGAMOS UNA COLUMNA NUEVA CON LAS ETIQUETAS REALES
frame['nombre_etiqueta']= iris.target_names[iris.target] #AGREGAMOS UNA COLUMNA CON LOS NOMBRES DE ETIQUETA
print(frame[frame['etiqueta']==1]) #EJEM. DE FILTRO CON PANDAS (VIRGÍNICA)
#print(frame) #MOSTRAMOS TODA LA DATA
```

```python
#DIVIDIMOS EL DATASET EN 2 PARTES: ENTRENAMIENTO (80%) Y PRUEBA (20%)
x_train, x_test, y_train, y_test= train_test_split(x, y, test_size=0.2, random_state= 42)
```

```python
#CREACIÓN Y ENTRENAMIENTO DEL MODELO (ÁRBOL DE DECISIONES PARA CLASIFICACIÓN MULTICLASE)
model= DecisionTreeClassifier()
model.fit(x_train, y_train)
```

```python
#PREDICCIÓN INDIVIDUAL
print('Predicción individual')
muestra_prueba= np.array([[6.4,3.2,4.5,1.5]]) #EXTRAEMOS UNA MUESTRA INDIVIDUAL
prediccion= model.predict(muestra_prueba) #OBTENEMOS LA ETIQUETA REAL DE LA PREDICCIÓN
print(prediccion)
print(iris.target_names[prediccion]) #OBTENEMOS EL NOMBRE DE LA ETIQUETA OBTENIDA
```
