```python
import torch
from transformers import pipeline

print("Torch disponible:", torch.__version__)
```

```python
analizador = pipeline(
    "sentiment-analysis",
    model="pysentimiento/robertuito-sentiment-analysis",
    device=-1  # -1 = CPU. Si tienes GPU con CUDA, puedes usar device=0
)
```

### Prueba del modelo con textos de ejemplo

```python
textos = [
    "Me encanta este producto",
    "Es horrible, no lo recomiendo",
    "Excelente atención al cliente",
    "La experiencia fue muy mala",
    "Estoy muy satisfecho con el servicio"
]

resultados = analizador(textos)
for texto, resultado in zip(textos, resultados):
    print(f"{texto!r} -> {resultado['label']} ({resultado['score']:.2f})")
```
